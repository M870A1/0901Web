```python
#if문
if(True):
    #실행문의 경우 들여쓰기 모두 같아야 정상적으로 실행됨
    print("조건식이 True인 경우 실행")
    print("두번째 실행문")
    # 들여쓰기를 하지 않으면 if문이 끝난것으로 설정됨
print("if문이 끝났습니다.")

#pass : if문 안에 실행문이 없이 실행하도록 만드는 명령어
if(True):
    pass
print("if문 종료")
```

```python
#if, else, else
a = 10
if(a>=10):
 print("a는 10보다 크거나 같습니다.")
elif(a<10): # 위 조건식이 해당 하지 않을시 실행
  print("a는 10보다 작습니다.")
  print("두번째 실행문")
else: # 위 모든 조건식이 해당 하지 않을시 실행
    print("a는 위의 두가지에 해당하지 않습니다.")
```

```python
# 논리 연산자
# >, >=, <, <=, ==, !=, and, or
print("10!=10",10!=10)
print("10=10",10==10)
print("15>10",15>10)
print("15>=10",15>=10)
print("15<15",15<15)
print("15<=15",15<=15)
print("3>5 and 5>3",3>5 and 5>3)
print("3>5 or 5>3",3>5 or 5>3)


```
