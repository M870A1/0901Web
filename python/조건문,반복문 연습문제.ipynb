```python
# 0 : 구구단 2~9단까지 출력하기
for dan in range(2, 10):  
    for i in range(1, 10):  
        print(f'{dan} x {i} = {dan * i}')
        #변수 dan * 변수i {} < 안에 변수를 바로 대입시키기 위해 사용
```

```python
# 1 : while문을 사용해 1부터 1000까지의 자연수 중 3의 배수의 합을 구해 보자.
a = 0
i = 1

while i <= 1000:
    if i % 3 == 0:
        a = a + i 
    i = i + 1 

print(a)
```

```python
# 3 : 1~100 중 홀수만을 저장하는 리스트를 만들어 보자
a = 0 # 초기값
while (True): #조건식
    a+=1 #증감식
    if(a>100):
        break # 반복문 강제 종료
    if(a%2==0):
        continue # 다음 반복으로 건너뛰기
    print("저장 : ",a)
```

```python
# 4 : 리스트에서 글자 수가 세글자 이상인 문자만 출력해보자
a = ['Life','is','too','short',',','you','need','python']
for i in a:
    if len(i) >= 3:  
        print(i)


#인터넷 찾아보니 밑에 식이 나옴
filtered_list = [word for word in a if len(word) >= 3]

print(filtered_list)

filtered_list = list(filter(lambda word: len(word) >= 3, a))
print(filtered_list)
```

```python
# 5 : 1900~2023년 까지의 윤년을 출력해보자
# 윤년의 조건은 4로 나누어 떨어지고,100으로 나누어떨어지지 않거나,400으로 나누어 떨어지는해
for year in range(1900 , 2023+1):
   if (year % 4 == 0 and year % 100 != 0 or year % 400 == 0):
     print(year)
```

```python
# 6 : 가위바위보 게임에서 승리 횟수가 5회 이상일때 게임이 끝나도록 만들어 보자
# 출력결과 : '승리 : 0/5', 패배 : 1, 비김 : 0'
#'승리 : 1/5', 패배 : 1, 비김 : 0'
#'승리 : 2/5', 패배 : 1, 비김 : 0'
#'승리 : 3/5', 패배 : 1, 비김 : 0'
#'승리 : 3/5', 패배 : 2, 비김 : 0'
#'승리 : 4/5', 패배 : 2, 비김 : 0'
#'승리 : 5/5', 패배 : 2, 비김 : 0'
#'다섯번 이겼습니다. 축하합니다.'
import random
data = ['가위', '바위', '보']

win = 0
lose = 0
draw = 0

while win < 5:
    # input(설명문) : 데이터를 입력하는 함수
    inputData = input("가위, 바위, 보 중에 입력해주세요 >> ")
	# random.choice(리스트) : 리스트중에 하나를 랜덤으로 출력하는 함수
    randomValue = random.choice(data)
    print(f"컴퓨터 : {randomValue}")
    print(f"플레이어 : {inputData}")

    if inputData == randomValue:
        draw += 1
        print("비겼습니다.")
    elif (inputData == '가위' and randomValue == '보' or 
         inputData == '바위' and randomValue == '가위' or
         inputData == '보' and randomValue == '바위'):
        win += 1
        print("승리했습니다!")
    else:
        lose += 1
        print("졌습니다.")
        
    # f-string 출력방식 : 문자열 안에 변수를 적기 쉽도록 하는 출력방식
    print(f"승리 : {win}/5, 패배 : {lose}, 비김 : {draw}")
print("다섯 번 이겼습니다. 축하합니다!")
```
